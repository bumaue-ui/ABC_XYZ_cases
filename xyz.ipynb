```python
import pandas as pd
import numpy as np
```

```python
pd.set_option('display.max_columns',40)
pd.set_option('display.float_format','{:2f}'.format)
```

```python
df = pd.read_csv('initial_data.csv', encoding='utf-8')

our_columns = ['DR_Dat','DR_NDrugs','DR_Kol']
df = df[our_columns]
df.columns = ['dt','name','quantity']
df['quantity'] = df['quantity'].replace({'\xa0': '', ',': '.'}, regex=True).astype(float)
```

```python
xyz_pivot = (df
         .pivot_table(index='name',
                      columns='dt',
                      values='quantity',
                      aggfunc='sum',
                      fill_value=0))

xyz_pivot
```

```python
mean = xyz_pivot.mean(axis=1)
std  = xyz_pivot.std(axis=1, ddof=0)
```

```python
xyz_analysis = (pd.DataFrame({'mean': mean, 'std': std})
              .assign(CV=lambda d: d['std'] / d['mean'],
                      XYZ=lambda d: pd.cut(d['CV'],
                                           bins=[-np.inf, 0.5, 1.0, np.inf],
                                           labels=['X', 'Y', 'Z'])))
```
